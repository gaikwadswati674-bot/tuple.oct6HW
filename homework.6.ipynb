```python
total_students = ["raj", "Amit", "Jay", "Payal", "Priya", "rahul", "Pavan", "Kiran", "Prachi"]
present_attendence = ("raj", "jay", "payal", "priya", "pavan", "kiran")

present = {name.lower() for name in present_attendence}
absent_students = [
    student for student in total_students
    if student.lower() not in present
]

print(", ".join(absent_students), "are absent today")
print("Total absent students:", len(absent_students))
```
